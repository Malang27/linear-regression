# Exercise 1: Scikit-learn estimator

#### 1.Fit a LinearRegression

```python
X, y = [[1],[2.1],[3]], [[1],[2],[3]]
model = LinearRegression()
model.fit(X, y)
x_pred = [[4]]
model.predict(x_pred)
```

# Exercise 5: Gradient Descent (Optional)

#### Question 1: Plot the data using

```python
X, y, coef = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=10,
                         coef=True,
                         random_state=0,
                         bias=100.0)
model = LinearRegression()
model.fit(X,y)
plt.scatter(X, y, color='blue', label='Data points')
plt.title(f"Generated Data with Fitted Linear Regression Line")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
def compute_mse(coefs, X, y):
    a, b = coefs
    y_preds = a * X.flatten() + b  # Ensure y_preds and y have the same shape
    mse = mean_squared_error(y, y_preds)
    
    return mse

# Coefficients a=1 and b=2
coefs = [1, 2]

# Compute the MSE
mse = compute_mse(coefs, X, y)
print("Mean Squared Error:", mse)
```

```python
aa, bb = np.mgrid[-200:200:0.5, -200:200:0.5]
grid = np.c_[aa.ravel(), bb.ravel()]
grid.shape
```

```python
from functools import partial
import multiprocessing


partial_compute_mse = partial(compute_mse, X=X, y=y)

# Parallelize the MSE computations
with multiprocessing.Pool() as pool:
    losses = pool.map(partial_compute_mse, grid)

# Convert the losses to a numpy array and reshape to match the grid shape
losses = np.array(losses).reshape(aa.shape)
```

```python
print(losses[0][:10])
```

```python
aa, bb = np.mgrid[-200:200:.5, -200:200:.5]
grid = np.c_[aa.ravel(), bb.ravel()]
losses_reshaped = np.array(losses).reshape(aa.shape)

f, ax = plt.subplots(figsize=(8, 6))
contour = ax.contourf(aa,
                    bb,
                    losses_reshaped,
                    100,
                    cmap="RdBu",
                    vmin=0,
                    vmax=160000)
ax_c = f.colorbar(contour)
ax_c.set_label("MSE")

ax.set(aspect="equal",
    xlim=(-200, 200),
    ylim=(-200, 200),
    xlabel="$a$",
    ylabel="$b$")
```

```python
min_index = np.argmin(losses)
optimal_a, optimal_b = grid[min_index]
print(optimal_a)
# Print the optimal coefficients
print(f"Optimal a: {optimal_a}, Optimal b: {optimal_b}")

# Plot the original data and the best fit line
plt.scatter(X, y, color='blue', label='Data')
y_optimal_preds = optimal_a * X.flatten() + optimal_b
plt.plot(X, y_optimal_preds, color='red', label='Best Fit Line')
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
a, b = 0, 0
num_iterations =100
learning_rate = 0.1
history = np.zeros((num_iterations, 2))
X =X.reshape(1,-1)
for i in range(num_iterations):
    y_pred = a * X + b
    
    da = ((y_pred- y) * X).mean()*2
    db = (y_pred-y).mean()*2
    
    a =a- learning_rate * da
    b =b - learning_rate * db
    history[i] = [a,b]

a,b
```

```python
aa, bb = np.mgrid[-200:200:.5, -200:200:.5]
grid = np.c_[aa.ravel(), bb.ravel()]
losses_reshaped = np.array(losses).reshape(aa.shape)

f, ax = plt.subplots(figsize=(8, 6))
contour = ax.contourf(aa,
                    bb,
                    losses_reshaped,
                    100,
                    cmap="RdBu",
                    vmin=0,
                    vmax=160000)
ax_c = f.colorbar(contour)
ax_c.set_label("MSE")

ax.set(aspect="equal",
    xlim=(-200, 200),
    ylim=(-200, 200),
    xlabel="$a$",
    ylabel="$b$")
ax.scatter(history[:, 0], history[:, 1])
    
    # Highlight start and end points (also in blue)
plt.show()
    
```
